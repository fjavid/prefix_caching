```python
from prompt_generator import (
    PromptGenerator,
    RAGExample,
    ScientificExample,
)

gen = PromptGenerator(seed=42)

rag_base = RAGExample(
    system_instruction="You are a helpful QA assistant. Use the retrieved context.",
    user_query="What are the main causes of coastal erosion?",
    retrieved_chunks=[
        "Coastal erosion is often driven by wave action, storms, and sea-level rise.",
        "Human activities such as dredging and construction can accelerate shoreline loss.",
        "Vegetation and wetlands can reduce erosion by stabilizing sediment."
    ],
    output_instruction="Answer in a short paragraph."
)

record = gen.generate_rag_pair(
    base=rag_base,
    mutation_type="document_replacement",
    candidate_chunks=[
        "Sediment transport patterns strongly affect long-term shoreline shape."
    ],
)

print(record.base_prompt)
print("-----")
print(record.mutated_prompt)

sci_base = ScientificExample(
    problem_description="Simulate 1D heat diffusion in a rod.",
    equation_name="Heat equation",
    parameters={"alpha": 0.1, "rod_length": 1.0},
    grid={"nx": 64, "nt": 100},
    constraints={"boundary": "u(0,t)=0, u(L,t)=0", "initial": "sin(pi x)"},
    output_schema='Return JSON with keys: "time" and "state".'
)

record = gen.generate_scientific_pair(
    base=sci_base,
    mutation_type="parameter_change",
    mutation_severity=1.0,
)

print(record.base_prompt)
print("-----")
print(record.mutated_prompt)
```
